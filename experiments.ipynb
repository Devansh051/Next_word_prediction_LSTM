```python
## Data Collection
import nltk
nltk.download('gutenberg')
from nltk.corpus import gutenberg
import pandas as pd
```

```python
## load the dataset
data = gutenberg.raw('shakespeare-hamlet.txt')

## save to a file
with open('hamlet.txt', 'w') as file:
    file.write(data)
```

```python
## Data Preprocessing
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences
from sklearn.model_selection import train_test_split

## load the dataset 
with open('hamlet.txt', 'r') as file:
    text = file.readlines() ## give a list of text.

## creating the indexes for the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts(text) # requires list of sentences.

total_words = len(tokenizer.word_index) + 1
total_words
```

```python
text
```

```python
tokenizer.word_index
```

- We have done this 'tokenizer.texts_to_sequences([line])[0]' because the text_to_sequences always returns a list of sequences because it expects a list of texts. so it gives list of list like this [[1, 687, 4, 45]] that's why [0]


- And since blank line or "/n" were giving [] in our sequence so to avoid we have used this if condition

```python
## creating input sequences for the sentence
## coverting our text into sequences. I love AI => [2,6,7] like that
input_sequences = []
for line in text:
    
    token_list = tokenizer.texts_to_sequences([line])[0] ## it should be in list form 
    if len(token_list)==0:
        continue

    ## so whenever we are doing this next_word prediction model then we create n_gram sequence.
    ## It is standard approach because each n-gram give the model training example.
    for i in range(1, len(token_list)):
        n_grams_sequence = token_list[:i+1]
        input_sequences.append(n_grams_sequence)

input_sequences
```

```python
## Pad_sequences
max_sequence_len = max([len(x) for x in input_sequences])
max_sequence_len
```

```python
input_sequences = pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')
input_sequences
```

```python
## create predictors and labels
import tensorflow as tf
x,y=input_sequences[:, :-1], input_sequences[:, -1]
x
```

```python
y
```

```python
## converting i/p label into categories using one hot encoding
y = tf.keras.utils.to_categorical(y, num_classes=total_words)
y
```

```python
# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

- return_sequences is important for stacking when use multiple LSTM layer because it return the hidden state every time step which is expected by second LSTM as input

```python
## Train our LSTM
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, LSTM, Dropout
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV

loss = tf.keras.losses.CategoricalCrossentropy()

## Define the model

model = Sequential()
model.add(Embedding(total_words, output_dim=100, input_length=max_sequence_len - 1)) ## since our x_train is 13 length as we have removed last col for y_train
model.add(LSTM(150, return_sequences=True))
model.add(Dropout(0.2))
model.add(LSTM(100)) ## other layer
model.add(Dense(total_words, activation="softmax"))
model.add(Dropout(0.3))
model.compile(optimizer="adam", loss = loss, metrics=["accuracy"])
  
```

```python
from tensorflow.keras.callbacks import TensorBoard,EarlyStopping
log_dir = 'logs/fit'
tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1, write_graph=True)
```

```python
earlyStopping_callback = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
## convert the keras_model to the scikit estimator
history = model.fit(x_train, y_train, verbose=1, epochs=200, batch_size=64, validation_split=0.2, callbacks=[tensorboard_callback])
```

- If the user enters more words than the model can accept, keep only the last (max_sequence_len - 1) words, because those are the most relevant for predicting the next word.

```python
model.save("next_word_lstm.h5")
```

```python
## function to predict the next word
def predict_next_word(model, tokenizer, text, max_sequence_len):
    token_list = tokenizer.texts_to_sequences([text])[0]

    ## this make sure if new sentence > max then it selects max_len_words from last 
    ## since to predict next_word context from last word play more role.
    if len(token_list) > max_sequence_len:
        token_list = token_list[-(max_sequence_len - 1):]

    token_list = pad_sequences([token_list], maxlen=max_sequence_len - 1, padding='pre')
    predicted = model.predict(token_list, verbose=0)

    predicted_word_index = np.argmax(predicted, axis=1)[0]
    for word, index in tokenizer.word_index.items():
        if index == predicted_word_index:
            return word

    return None
```

```python
input_text="To be or not to be"
print(f"Input text:{input_text}")
max_sequence_len=model.input_shape[1]+1
next_word=predict_next_word(model, tokenizer, input_text, max_sequence_len)
print(f"Next word Prediction: {next_word}")
```
